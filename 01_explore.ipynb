```python
%sql
-- No repeated journey/rental IDs

SELECT `Rental Id`
FROM jan2016
GROUP BY `Rental Id`
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT *
FROM jan2016
ORDER BY `Rental Id`
LIMIT 20;
```

```python
%sql

SELECT 
    MIN(`End Date`),
    MAX(`End Date`),
    MIN(`Duration`),
    MAX(`Duration`)
FROM jan2016;
```

```python
%sql
WITH converted_dates AS (
    SELECT 
        TO_DATE(`End Date`, 'dd/MM/yyyy HH:mm') AS EndDate
    FROM jan2016
)

SELECT
    EndDate,
    COUNT(*) AS `Total Hires`
FROM converted_dates
GROUP BY EndDate
ORDER BY EndDate;
```
